```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.datasets import load_iris
```

```python
iris = load_iris()

df = pd.DataFrame(iris.data, columns=iris.feature_names)
df["Target"] = iris.target

df.head()
```

```python
df.describe()
```

### **Task 1:** Feature-wise Statistical Properties

#### MEAN

```python
df.mean(numeric_only=True)
```

#### MEADIAN

```python
df.median(numeric_only=True) 
```

#### MODE

```python
df.mode()
```

#### Standard Deviation

```python
df.std(numeric_only=True)
```

#### Variance

```python
df.var(numeric_only=True)
```

#### Skewness

```python
df.skew(numeric_only=True)
```

#### Kurtosis

```python
df.kurt(numeric_only=True)
```

### **Task 2:** Class-wise Statistical Properties

#### MEAN

```python
df.groupby("Target").mean()
```

#### MEDIAN

```python
df.groupby("Target").median()
```

#### MODE

```python
df.groupby("Target").agg(lambda x: x.mode().iloc[0])
```

#### Standard Deviation

```python
df.groupby("Target").std()
```

#### Variance

```python
df.groupby("Target").var()
```

#### Skewness

```python
df.groupby("Target").skew()
```

#### Kurtosis

```python
df.groupby("Target").apply(pd.DataFrame.kurt)
```

### **Task 3:** Correlation Between Features

```python
corr = df.drop("Target", axis=1).corr()
corr
```

### **Task 4:** Graph

#### Heatmap

```python
plt.figure(figsize=(8,6))
sns.heatmap(corr, annot=True, cmap="coolwarm")
plt.title("Correlation Matrix")
plt.show()
```

#### Box Plot (Quartile Plot)

```python
plt.figure(figsize=(10,6))
sns.boxplot(data=df.drop("Target", axis=1))
plt.xticks(rotation=45)
plt.show()
```

#### Scatter Plot

```python
plt.figure(figsize=(8,6))
sns.scatterplot(
    data=df,
    x="sepal length (cm)",
    y="petal length (cm)",
    hue="Target"
)
plt.show()
```
